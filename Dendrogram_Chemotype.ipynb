```python
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

# === Step 1: Load and preprocess ===
file_path = r"C:\Users\exm499\Downloads\PFAS_classification\Chemotype\Chemotype_15PFAS.xlsx"
df_chem = pd.read_excel(file_path, header=4)
df_t = df_chem.set_index("M_NAME").T
df_t.index.name = "CompoundID"

# === Step 2: Hierarchical clustering ===
linkage_matrix = linkage(df_t, method="ward", metric="euclidean")

# === Step 3: Create PFAS1 ~ PFASn labels ===
compound_labels = [f"PFAS{i+1}" for i in range(len(df_t))]

# === Step 4: Plot dendrogram ===
plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, labels=compound_labels, leaf_rotation=90)
plt.title("Dendrogram of PFAS Compounds Based on Chemotype (PFAS1, PFAS2...)")
plt.xlabel("PFAS Compounds")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```
