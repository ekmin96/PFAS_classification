```python
import pandas as pd
import os
import re

input_path = r"C:\Users\meg57\OneDrive\바탕 화면\은기\진행중\PFAS_transcriptomics_read-across\PFAS_matrix\PFAS_genes_expected_counts_Filtered.xlsx"
output_dir = r"C:\Users\meg57\OneDrive\바탕 화면\은기\진행중\PFAS_transcriptomics_read-across\PFAS_matrix\Raw-files"
os.makedirs(output_dir, exist_ok=True)

comparison_dict = {
    "PFAS1":  "ControlA",
    "PFAS2":  "ControlB",
    "PFAS3":  "ControlA",
    "PFAS4":  "ControlB",
    "PFAS5":  "ControlA",
    "PFAS6":  "ControlA",
    "PFAS7":  "ControlA",
    "PFAS8":  "ControlA",
    "PFAS9":  "ControlA",
    "PFAS10": "ControlA",
    "PFAS11": "ControlA",
    "PFAS12": "ControlA",
    "PFAS13": "ControlA",
    "PFAS14": "ControlB",
    "PFAS15": "ControlB"
}

df = pd.read_excel(input_path)
df = df.set_index(df.columns[0])  # GeneID를 인덱스로
df.insert(0, "description", "NA")  # description 컬럼 추가 (모두 NA로)

all_columns = df.columns.tolist()

for pfas, control in comparison_dict.items():
    control_cols = [col for col in all_columns if re.match(f"{control}_\\d+", col)]
    pfas_cols = [col for col in all_columns if re.match(f"{pfas}_\\d+", col)]
    
    cols_to_save = ["description"] + control_cols + pfas_cols
    df_sub = df[cols_to_save].copy()
    
    df_sub.insert(0, "GeneID", df_sub.index)
    
    out_file = os.path.join(output_dir, f"{pfas}_vs_{control}.txt")
    df_sub.to_csv(out_file, sep="\t", index=False)
    
    print(f"✅ 저장 완료: {out_file}")
```

```python
import pandas as pd
import os
import glob

input_dir = r"C:\Users\meg57\OneDrive\바탕 화면\은기\진행중\PFAS_transcriptomics_read-across\PFAS_matrix\DEseq_FC"
output_dir = r"C:\Users\meg57\OneDrive\바탕 화면\은기\진행중\PFAS_transcriptomics_read-across\PFAS_matrix\Ranking"
os.makedirs(output_dir, exist_ok=True)

csv_files = glob.glob(os.path.join(input_dir, "DEG_PFAS*_vs_*.csv"))

if not csv_files:
    print("❌ '.csv' 파일을 찾지 못했습니다.")
    exit()

for file_path in csv_files:
    file_name = os.path.basename(file_path)
    base_name = os.path.splitext(file_name)[0]  # "DEG_PFAS1_vs_ControlA"
    pfas_name = base_name.replace("DEG_", "")   # "PFAS1_vs_ControlA"

    df = pd.read_csv(file_path, index_col=0)

    df = df.dropna(subset=["log2FoldChange", "lfcSE"])
    df["stat"] = df["log2FoldChange"] / df["lfcSE"]

    df_stat = df[["stat"]].sort_values(by="stat", ascending=False)

    output_file = os.path.join(output_dir, f"{pfas_name}.rnk")
    df_stat.to_csv(output_file, sep="\t", header=False, index=True)

    print(f"✅ 저장 완료: {output_file}")
```

```python
import gseapy as gp
import os

base_path = r"C:\Users\meg57\OneDrive\바탕 화면\은기\진행중\PFAS_transcriptomics_read-across\PFAS_matrix"
rnk_dir = os.path.join(base_path, "Ranking")
gmt_file = os.path.join(base_path, "gmt", "zebrafish_GO_KEGG_EnsemblID_merged.gmt")  # <- 원하는 gmt 파일명으로 교체
output_base = os.path.join(base_path, "GSEA_output")
os.makedirs(output_base, exist_ok=True)

batch_A = [f"PFAS{i}" for i in [1, 3] + list(range(5, 14))]
batch_B = [f"PFAS{i}" for i in [4, 14, 15]]
batch_2 = ["PFAS2"]

all_batches = batch_A + batch_B + batch_2

for pfas in all_batches:
    # .rnk 파일 찾기
    rnk_file = os.path.join(rnk_dir, f"{pfas}_vs_ControlA.rnk") if pfas in batch_A else os.path.join(rnk_dir, f"{pfas}_vs_ControlB.rnk")

    output_dir = os.path.join(output_base, pfas)
    os.makedirs(output_dir, exist_ok=True)

    try:
        gp.prerank(
            rnk=rnk_file,
            gene_sets=gmt_file,
            outdir=output_dir,
            permutation_num=1000,
            min_size=15,
            max_size=500,
            format='png',
            seed=42,
            verbose=True
        )
        print(f"✅ GSEA 완료: {pfas}")
    except Exception as e:
        print(f"❌ GSEA 실패: {pfas} → {e}")
```
